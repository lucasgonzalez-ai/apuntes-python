## 1. ¿Qué es la POO?

### 📖 Marco Teórico

La **Programación Orientada a Objetos (POO)** es un paradigma de programación que organiza el software en torno a **objetos** en lugar de funciones y lógica. Un objeto es una entidad que combina:

- **Datos** (atributos)
- **Comportamientos** (métodos)

### 🌍 Analogía con el mundo real

Pensemos en un **automóvil**:

| Concepto POO | Ejemplo Auto |
|---|---|
| Clase | El plano/diseño del auto |
| Objeto | Un auto específico (tu Toyota Corolla rojo) |
| Atributo | Color, marca, velocidad, combustible |
| Método | Acelerar, frenar, encender, apagar |

### 🏛️ Los 4 Pilares de la POO

```
┌─────────────────────────────────────────┐
│           PILARES DE LA POO             │
│                                         │
│  🔒 Encapsulamiento  🧬 Herencia        │
│  🎭 Polimorfismo     🎨 Abstracción     │
└─────────────────────────────────────────┘
```

### ✅ Ventajas de la POO

- **Reutilización de código**: Una clase puede usarse en múltiples proyectos
- **Modularidad**: El código se divide en partes independientes
- **Mantenibilidad**: Es más fácil actualizar partes sin romper el todo
- **Modelado real**: El código refleja cómo pensamos el mundo

## 2. Clases y Objetos

### 📖 Concepto

- **Clase**: Es el *molde* o *plantilla* que define la estructura y comportamiento de un tipo de objeto. Es abstracta por sí misma.
- **Objeto (Instancia)**: Es una *realización concreta* de una clase. Cada objeto tiene sus propios valores de atributos.

### 🔑 Sintaxis básica en Python

```python
class NombreClase:       # Definición de la clase
    def __init__(self):  # Constructor (método especial)
        pass             # Cuerpo del constructor

objeto = NombreClase()   # Creación del objeto (instanciación)
```

### 📌 El método `__init__` (Constructor)

Es el método que Python llama **automáticamente** cuando se crea un objeto. Se usa para inicializar los atributos. El parámetro `self` representa al objeto que se está creando.

````python
# ============================================================
# EJEMPLO 1: Primera clase en Python
# ============================================================

class silla:
    nombre = 'silla1'
    color = 'verde'
    material = 'madera'

s=silla()
print(s.nombre) # Acceso a un atributo de clase
print(s.color)
print(s.material)

p=silla()
print(p.nombre) # Acceso a un atributo de clase
print(p.color)
print(p.material)

class Perro:
    """
    Clase que representa a un perro.

    Los triple comillas se llaman 'docstring' y sirven
    para documentar la clase.
    """

    nivel = 1 # Atributo de clase que indica el nivel de la clase Perro (puede ser usado para clasificar o categorizar perros)

    def __init__(self, nombre, raza, edad, color, esta_vivo=True):
        """
        Constructor: se ejecuta al crear un nuevo Perro.

        Parámetros:
            nombre (str): El nombre del perro
            raza   (str): La raza del perro
            edad   (int): La edad en años
        """
        # 'self.nombre' crea un atributo de instancia
        # El lado izquierdo (self.nombre) es el atributo del objeto
        # El lado derecho (nombre) es el parámetro recibido
        self.nombre = nombre
        self.raza   = raza
        self.edad   = edad
        self.color = color # Atributo adicional con valor por defecto
        self.esta_vivo = esta_vivo # Atributo adicional para indicar si el perro está vivo (valor inicial: True)

    def ladrar(self, veces): # sin parámetros adicionales a 'self'
        """Método que hace ladrar al perro."""
        print(f"{self.nombre} dice: ¡Guau guau! ( {veces} veces )") # 'veces' es un parámetro adicional que indica cuántas veces ladrar

    def presentarse(self):
        """Método que muestra la información del perro."""
        print(f"Hola, soy {self.nombre}, un {self.raza} de {self.edad} año(s) y de color {self.color}.")
        # la funcion f-string permite insertar variables dentro de una cadena de texto con el formato {variable}

    def sentarse(self):
        """Método que hace sentar al perro."""
        print(f"{self.nombre} se sienta.")

    def dar_la_pata(self):
        """Método que hace dar la pata al perro."""
        print(f"{self.nombre} da la pata.")

    def cumplir_años(self):
        """Método que incrementa la edad del perro en 1 año."""
        self.edad += 1
        print(f"{self.nombre} ha cumplido años y ahora tiene {self.edad} año(s).")

    def vivo_muerto(self):
        """Método que muestra si el perro está vivo o muerto."""
        estado = "vivo" if self.esta_vivo else "muerto"
        print(f"{self.nombre} está {estado}.")
# --- Creación de objetos (instanciación) ---
# Cada objeto es independiente y tiene sus propios valores

p1 = Perro("Rex","Pastor Alemán",3, 'Negro')
p2 = Perro("Luna",  "Labrador",5, 'Blanco')
p3 = Perro("Toby",  "Chihuahua",1, 'Café')

````

````python
print(p1.nombre)
print(p1.color)

p1.color = 'negro'
p1.nombre = 'TRex'

print(p1.nombre)
print(p1.color)


# --- Usando los objetos ---
print("=" * 40) # Imprime una línea de separación 40 veces el signo =
print("Presentaciones:")
print("=" * 40)

p1.presentarse()
p2.presentarse()
p3.presentarse()

p1.dar_la_pata()

print("\nLadridos:")
""" \n es un salto de línea para separar secciones se usa al principio de la cadena para crear un espacio
antes del texto o al final para crear un espacio después del texto """
p1.ladrar(5)
p2.ladrar(10)

# --- Acceso directo a atributos ---
print(f"\nEdad de {p1.nombre}: {p1.edad} años") # objeto.atributo para acceder al valor del atributo
print(f"Raza de {p2.nombre}: {p2.raza}")
````

## 3. Atributos y Métodos

### 📖 Tipos de Atributos

| Tipo | Descripción | Dónde se define |
|------|-------------|------------------|
| **de Instancia** | Pertenecen a cada objeto individual | Dentro de `__init__` con `self.` |
| **de Clase** | Compartidos por TODOS los objetos | Fuera de cualquier método, dentro de la clase |

### 📖 Tipos de Métodos

| Tipo | Decorador | Primer Parámetro | Accede a |
|------|-----------|-----------------|----------|
| **de Instancia** | ninguno | `self` | Atributos del objeto |
| **de Clase** | `@classmethod` | `cls` | Atributos de clase |
| **Estático** | `@staticmethod` | ninguno | Nada del objeto/clase |

````python
# ============================================================
# EJEMPLO 2: Tipos de atributos y métodos
# ============================================================

class CuentaBancaria:
    """
    Ejemplo que demuestra los distintos tipos de
    atributos y métodos en Python.
    """

    # ── ATRIBUTO DE CLASE ──────────────────────────────────
    # Compartido por TODAS las instancias de CuentaBancaria
    tasa_interes    = 0.05   # 5% anual para todos
    total_cuentas   = 0      # Contador global de cuentas

    def __init__(self, titular, saldo_inicial=0):
        # ── ATRIBUTOS DE INSTANCIA ─────────────────────────
        # Únicos para cada objeto
        self.titular  = titular
        self.saldo    = saldo_inicial
        self.numero   = CuentaBancaria.total_cuentas + 1

        # Incrementamos el contador de clase
        CuentaBancaria.total_cuentas += 1

    # ── MÉTODO DE INSTANCIA ────────────────────────────────
    # Opera sobre el objeto específico (usa self)
    def depositar(self, monto):
        """Agrega dinero a la cuenta."""
        if monto > 0:
            self.saldo += monto
            print(f"✅ Depósito de ${monto:,.2f}. Nuevo saldo: ${self.saldo:,.2f}")
        else:
            print("❌ El monto debe ser positivo.")

    def retirar(self, monto):
        """Retira dinero de la cuenta."""
        if monto <= self.saldo:
            self.saldo -= monto
            print(f"💸 Retiro de ${monto:,.2f}. Nuevo saldo: ${self.saldo:,.2f}")
        else:
            print(f"❌ Fondos insuficientes. Saldo actual: ${self.saldo:,.2f}")

    def calcular_interes_anual(self):
        """Calcula el interés anual usando el atributo de clase."""
        # Accede tanto al atributo de instancia (self.saldo)
        # como al de clase (CuentaBancaria.tasa_interes)
        return self.saldo * CuentaBancaria.tasa_interes

    def mostrar_estado(self):
        """Muestra el estado de la cuenta."""
        print(f"\n{'='*35}")
        print(f"  Cuenta N°: {self.numero}")
        print(f"  Titular:   {self.titular}")
        print(f"  Saldo:     ${self.saldo:,.2f}")
        print(f"  Interés:   ${self.calcular_interes_anual():,.2f}")
        print(f"{'='*35}")

    # ── MÉTODO DE CLASE ────────────────────────────────────
    # Opera sobre la clase (no sobre un objeto específico)
    # Recibe 'cls' en lugar de 'self'
    @classmethod
    def cambiar_tasa(cls, nueva_tasa):
        """Cambia la tasa de interés para TODAS las cuentas."""
        cls.tasa_interes = nueva_tasa
        print(f"📊 Nueva tasa de interés global: {nueva_tasa*100:.1f}%")

    @classmethod
    def ver_total_cuentas(cls):
        """Muestra cuántas cuentas existen en total."""
        print(f"🏦 Total de cuentas en el banco: {cls.total_cuentas}")

    # ── MÉTODO ESTÁTICO ────────────────────────────────────
    # No accede a self ni a cls
    # Es una función de utilidad relacionada con la clase
    @staticmethod
    def es_monto_valido(monto):
        """Verifica si un monto es un número positivo válido."""
        return isinstance(monto, (int, float)) and monto > 0


# ── DEMO ──────────────────────────────────────────────────
print("Creando cuentas bancarias...\n")

cuenta_juan  = CuentaBancaria("Juan Pérez",   10000)
cuenta_maria = CuentaBancaria("María García",2500)

cuenta_juan.mostrar_estado()
cuenta_maria.mostrar_estado()

print("\nOperaciones:")
cuenta_juan.depositar(2500)
cuenta_juan.retirar(500)
cuenta_maria.retirar(9999)  # Intento fallido

# Método de clase: afecta a TODAS las cuentas
print("\nCambiando tasa de interés global:")
CuentaBancaria.cambiar_tasa(0.08)
print(f"Nuevo interés de Juan:  ${cuenta_juan.calcular_interes_anual():,.2f}") # formato :,.2f para mostrar el número con comas y 2 decimales
print(f"Nuevo interés de María: ${cuenta_maria.calcular_interes_anual():,.2f}")# llama a la función calcular_interes_anual para cada cuenta, que a su vez accede al atributo de clase tasa_interes actualizado

# Método de clase
CuentaBancaria.ver_total_cuentas()

# Método estático (se puede llamar sin instancia)
print(f"\n¿1500 es válido? {CuentaBancaria.es_monto_valido(1500)}")# llama a la función es_monto_valido sin necesidad de crear una cuenta, ya que es un método estático
print(f"¿-100 es válido? {CuentaBancaria.es_monto_valido(-100)}")
print(f"¿'abc' es válido? {CuentaBancaria.es_monto_valido('abc')}")
````

### 📖 Métodos Especiales (Dunder Methods)

Python tiene métodos con doble guión bajo (`__nombre__`) que permiten que los objetos interactúen con operadores y funciones integradas del lenguaje.

````python
# ============================================================
# EJEMPLO 3: Métodos especiales / Dunder methods
# ============================================================

class Vector2D:
    """Un vector matemático en 2 dimensiones."""

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self): # metodo especial sirve para mostrar el objeto de forma amigable al usuario, se llama automáticamente al usar print() o str() sobre el objeto
        """
        Define cómo se muestra el objeto con print().
        Orientado al usuario final.
        """
        return f"Vector({self.x}, {self.y})" #retorna una cadena con la representación del vector, usando los valores de x e y del objeto actual (self)

    def __repr__(self):# repr es una representación más detallada y técnica del objeto, útil para desarrolladores y debugging, se llama al usar repr() o al mostrar el objeto en un entorno interactivo
       #debugging, se llama al usar repr() o al mostrar el objeto en un entorno interactivo
       # debugging significa en español "depuración", es el proceso de identificar y corregir errores en el código
       # un entorno interactivo es aquel donde se muestra el resultado de cada expresión, como la consola de Python o un Jupyter Notebook
        """
        Representación técnica del objeto.
        Orientado a desarrolladores (debugging).
        """
        return f"Vector2D(x={self.x}, y={self.y})"

    def __add__(self, otro):# add es un método especial que permite definir el comportamiento del operador '+' para objetos de la clase Vector2D
        """
        Permite usar el operador '+' entre vectores.
        v1 + v2  →  llama a  v1.__add__(v2)
        """
        return Vector2D(self.x + otro.x, self.y + otro.y)

    def __sub__(self, otro):# sub es un método especial que permite definir el comportamiento del operador '-' para objetos de la clase Vector2D
        """Permite usar el operador '-' entre vectores."""
        return Vector2D(self.x - otro.x, self.y - otro.y)

    def __mul__(self, escalar):# mul es un método especial que permite definir el comportamiento del operador '*' para objetos de la clase Vector2D, en este caso multiplicar por un escalar (número)
        """Permite multiplicar un vector por un escalar."""
        return Vector2D(self.x * escalar, self.y * escalar)

    def __eq__(self, otro):# eq es un método especial que permite definir el comportamiento del operador '==' para objetos de la clase Vector2D, en este caso compara si dos vectores son iguales (tienen las mismas coordenadas)
        """Permite comparar con '=='."""
        return self.x == otro.x and self.y == otro.y

    def __len__(self):# len es un método especial que permite definir el comportamiento de la función len() para objetos de la clase Vector2D, en este caso retorna la cantidad de dimensiones del vector (2)
        """Permite usar len() – retorna dimensiones del vector."""
        return 2

    def magnitud(self):
        """Calcula la longitud del vector."""
        return (self.x**2 + self.y**2) ** 0.5


# Demo
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)

print(f"v1 = {v1}")              # Usa __str__
print(f"v2 = {v2}")
print(f"Repr: {repr(v1)}")       # Usa __repr__
print(f"v1 + v2 = {v1 + v2}")   # Usa __add__
print(f"v1 - v2 = {v1 - v2}")   # Usa __sub__
print(f"v1 * 3  = {v1 * 3}")    # Usa __mul__
print(f"v1 == v2: {v1 == v2}")  # Usa __eq__
print(f"len(v1): {len(v1)}")    # Usa __len__
print(f"|v1| = {v1.magnitud():.2f}")
````

## 4. Encapsulamiento

### 📖 Concepto

El **encapsulamiento** es el mecanismo de ocultar los detalles internos de un objeto y exponer solo lo necesario. Protege los datos de modificaciones accidentales o incorrectas.

### 🔐 Niveles de acceso en Python

Python usa convenciones de nomenclatura (no impone restricciones reales como Java/C++):

| Convención | Ejemplo | Significado |
|------------|---------|-------------|
| Sin guión bajo | `nombre` | Público – accesible desde cualquier lugar |
| Un guión bajo | `_nombre` | Protegido – convención: uso interno |
| Dos guiones bajos | `__nombre` | Privado – name mangling aplicado |

### 🔧 Properties (Getters y Setters)

Permiten controlar el acceso y la modificación de atributos con validación.

````python
# ============================================================
# EJEMPLO 4: Encapsulamiento y Properties
# ============================================================

class Persona:
    """
    Demuestra encapsulamiento con @property.
    """
    #@property es un decorador que convierte un método en una propiedad, lo que permite acceder a él como si fuera un atributo,
    # pero con la posibilidad de agregar lógica adicional (como validación) al obtener o establecer su valor. Esto es útil
    # para proteger atributos privados y controlar su acceso.

    # Getter es un método que se utiliza para obtener el valor de un atributo privado
    # Setter es un método que se utiliza para establecer o modificar el valor de ese atributo privado,
    # ambos se definen utilizando el decorador @property y @nombre.setter respectivamente, donde 'nombre' es el nombre del atributo que se desea proteger.

    def __init__(self, nombre, edad, email):
        # Atributo público
        self.nombre  = nombre

        # Atributo privado (name mangling: se convierte en _Persona__edad)
        # No se puede acceder directamente desde fuera
        self.__edad  = edad

        # Atributo protegido (convención: no acceder desde fuera)
        self._email  = email

    # ── PROPERTY (Getter) ──────────────────────────────────
    # Permite acceder a __edad como si fuera un atributo público
    # pero con control
    @property
    def edad(self):
        """Getter: obtiene la edad."""
        return self.__edad

    # ── SETTER ─────────────────────────────────────────────
    # Permite MODIFICAR __edad con validación
    @edad.setter
    def edad(self, nueva_edad):
        """Setter: modifica la edad con validación."""
        if not isinstance(nueva_edad, int):
            raise TypeError("La edad debe ser un número entero.")
        if nueva_edad < 0 or nueva_edad > 150:
            raise ValueError(f"Edad inválida: {nueva_edad}. Debe estar entre 0 y 150.")
        self.__edad = nueva_edad
        print(f"✅ Edad actualizada a {nueva_edad} años.")

    @property
    def email(self):
        """Getter del email."""
        return self._email

    @email.setter
    def email(self, nuevo_email):
        """Setter del email con validación básica."""
        if "@" not in nuevo_email or "." not in nuevo_email:
            raise ValueError(f"Email inválido: '{nuevo_email}'")
        self._email = nuevo_email
        print(f"✅ Email actualizado a '{nuevo_email}'.")

    def __str__(self):
        return f"Persona(nombre={self.nombre}, edad={self.__edad}, email={self._email})"


# ── DEMO ──────────────────────────────────────────────────
p = Persona("Carlos", 28, "carlos@email.com")
print(p)

# Acceso a través de property (se ve como atributo normal)
print(f"\nEdad: {p.edad}")
print(f"Email: {p.email}")

# Modificación con validación
print("\nModificaciones:")
p.edad = 30
p.email = "carlos.nuevo@gmail.com"

# Intentos de valores inválidos
print("\nIntentos inválidos:")
try:
    p.edad = -5
except ValueError as e:
    # ValueError es una clase de excepción que se lanza cuando se pasa un valor inapropiado a una función o método,
    # en este caso se lanza si la edad es negativa o mayor a 150
    print(f"❌ Error: {e}")

try:
    p.email = "correo-invalido"
except ValueError as e:
    print(f"❌ Error: {e}")

# Intentar acceder al atributo privado directamente
print("\nIntento de acceso directo al atributo privado:")
try:
    print(p.__edad)  # Esto falla
except AttributeError as e:
    print(f"❌ No se puede acceder: {e}")

# Pero con name mangling sí se puede (aunque no se debe)
print(f"Con name mangling (solo para aprendizaje): {p._Persona__edad}")
````

## 5. Herencia

### 📖 Concepto

La **herencia** permite crear una nueva clase basada en una existente. La clase nueva (*subclase* o *clase hija*) hereda los atributos y métodos de la clase original (*superclase* o *clase padre*).

```
        Animal (Superclase / Padre)
       /      \
    Perro    Gato   (Subclases / Hijas)
    /   \
Poodle Bulldog     (Subclases de Perro)
```

### 🔑 Sintaxis

```python
class ClaseHija(ClasePadre):
    def __init__(self, ...):
        super().__init__(...)  # Llama al constructor del padre
```

La función `super()` da acceso a los métodos de la clase padre.

````python
# ============================================================
# EJEMPLO 5: Herencia simple y múltiple niveles
# ============================================================

# ── CLASE BASE (Padre) ─────────────────────────────────────
class Animal:
    """Clase base para todos los animales."""

    def __init__(self, nombre, edad, peso):
        self.nombre = nombre
        self.edad   = edad
        self.peso   = peso
        self.vivo   = True

    def respirar(self):
        print(f"{self.nombre} está respirando.")

    def comer(self, alimento):
        print(f"{self.nombre} come {alimento}. ¡Qué rico!")

    def hacer_sonido(self):
        """Método genérico – las subclases lo sobreescribirán."""
        print(f"{self.nombre} hace un sonido.")

    def __str__(self):
        return f"{type(self).__name__}: {self.nombre} ({self.edad} años, {self.peso}kg)"


# ── SUBCLASE 1 ─────────────────────────────────────────────
class Perro(Animal):
    """Subclase de Animal para perros."""

    def __init__(self, nombre, edad, peso, raza):
        # super().__init__ llama al constructor del padre
        # Esto inicializa nombre, edad, peso y vivo
        super().__init__(nombre, edad, peso)

        # Atributo adicional específico de Perro
        self.raza = raza

    def hacer_sonido(self):  # SOBREESCRITURA del método padre
        print(f"{self.nombre} dice: ¡Guau! 🐕")

    def buscar_pelota(self):
        """Método exclusivo de Perro."""
        print(f"{self.nombre} busca la pelota con entusiasmo! 🎾")

    def __str__(self):
        base = super().__str__()  # Reutiliza el __str__ del padre
        return f"{base}, Raza: {self.raza}"


# ── SUBCLASE 2 ─────────────────────────────────────────────
class Gato(Animal):
    """Subclase de Animal para gatos."""

    def __init__(self, nombre, edad, peso, es_domestico=True):
        super().__init__(nombre, edad, peso)
        self.es_domestico = es_domestico

    def hacer_sonido(self):
        print(f"{self.nombre} dice: ¡Miau! 🐱")

    def ronronear(self):
        print(f"{self.nombre} ronronea felizmente... prrrrr 😺")


# ── SUBCLASE DE SUBCLASE (3 niveles) ──────────────────────
class PerroPolicia(Perro):
    """Subclase de Perro para perros policía."""

    def __init__(self, nombre, edad, peso, raza, numero_placa):
        super().__init__(nombre, edad, peso, raza)
        self.numero_placa = numero_placa

    def detectar(self):
        print(f"🚔 Oficial {self.nombre} (Placa: {self.numero_placa}) está detectando amenazas.")


# ── DEMO ──────────────────────────────────────────────────
print("Creando animales...\n")

perro     = Perro("Buddy",   3, 25.0,  "Golden Retriever")
gato      = Gato("Whiskers", 5, 4.5)
k9        = PerroPolicia("Rex", 4, 30.0, "Pastor Alemán", "K9-007")

print(perro)
print(gato)
print(k9)

print("\nSonidos:")
perro.hacer_sonido()
gato.hacer_sonido()
k9.hacer_sonido()

print("\nComportamientos heredados:")
perro.respirar()       # Heredado de Animal
gato.comer("atún")    # Heredado de Animal

print("\nComportamientos específicos:")
perro.buscar_pelota()  # Propio de Perro
gato.ronronear()       # Propio de Gato
k9.detectar()          # Propio de PerroPolicia

print("\nVerificación de herencia con isinstance():")
print(f"¿k9 es PerroPolicia? {isinstance(k9, PerroPolicia)}")  # True (herencia directa)
# isinstance() es una función incorporada en Python que se utiliza para verificar si un objeto
# es una instancia de una clase específica o de una clase derivada de ella, devuelve True si el objeto es una instancia de la clase
# o de sus subclases, y False en caso contrario.
print(f"¿k9 es Perro?        {isinstance(k9, Perro)}")         # True (herencia)
print(f"¿k9 es Animal?       {isinstance(k9, Animal)}")        # True (herencia transitiva)
print(f"¿k9 es Gato?         {isinstance(k9, Gato)}")          # False
````

### 📖 Herencia Múltiple

Python permite que una clase herede de más de una clase padre. Esto puede ser útil pero también complejo. Python usa el **MRO (Method Resolution Order)** para determinar el orden de búsqueda de métodos.

````python
# Python utiliza un algoritmo avanzado llamado C3 Linearization.
# Este algoritmo cumple dos reglas básicas:
# - Va de la clase más específica (la clase actual) hacia las más generales (los padres).
# - Respeta el orden en el que escribiste las clases padre de izquierda a derecha.
# Puedes consultarlo de dos formas sencillas en tu código:
__nombre_de_clase.__mro__
.mro()
````

````python
# ============================================================
# EJEMPLO 6: Herencia múltiple y MRO
# ============================================================

class Volador:
    def volar(self):
        print(f"{self.nombre} está volando 🦅")

    def habilidad_principal(self):
        return "volar"

class Nadador:
    def nadar(self):
        print(f"{self.nombre} está nadando 🐠")

    def habilidad_principal(self):
        return "nadar"


# Pato hereda de Animal, Volador Y Nadador
class Pato(Animal, Volador, Nadador):
    def __init__(self, nombre, edad, peso):
        super().__init__(nombre, edad, peso)

    def hacer_sonido(self):
        print(f"{self.nombre} dice: ¡Cuac cuac! 🦆")


donald = Pato("Donald", 3, 2.0)

donald.hacer_sonido()          # Propio
donald.volar()                  # De Volador
donald.nadar()                  # De Nadador
donald.comer("pan")            # De Animal

# MRO: el orden en que Python busca los métodos
print("\nMRO (Method Resolution Order):")
for i, clase in enumerate(Pato.__mro__):
    # enumerate es una función incorporada en Python que se utiliza para iterar sobre una secuencia (como una lista o tupla)
    # y obtener tanto el índice como el valor de cada elemento, devuelve un objeto enumerado que produce pares de índice y valor,
    # donde el índice comienza en 0 por defecto.

    # __mro__ es un atributo de clase que muestra el orden de búsqueda de métodos para la clase Pato,
    # es una tupla que contiene las clases en el orden en que se buscarán los métodos cuando se llamen sobre un objeto de esa clase
    print(f"  {i+1}. {clase.__name__}") # __name__ es un atributo que contiene el nombre de la clase como cadena de texto

# Como Animal va primero en el MRO, 'habilidad_principal' lo toma de Volador
# (Volador es el primero que tiene ese método después de Pato y Animal)
print(f"\nHabilidad principal: {donald.habilidad_principal()}")
````

## 6. Polimorfismo

### 📖 Concepto

El **polimorfismo** (del griego: *poli* = muchos, *morphos* = formas) permite que objetos de diferentes clases respondan al mismo mensaje (método) de maneras distintas.

**Principio**: Mismo nombre de método → comportamientos diferentes según el objeto.

```
figura.area()  →  círculo calcula πr²
figura.area()  →  rectángulo calcula largo × ancho
figura.area()  →  triángulo calcula base × altura / 2
```

````python
# ============================================================
# EJEMPLO 7: Polimorfismo con figuras geométricas
# ============================================================

import math
# math es un módulo de la biblioteca estándar de Python que proporciona funciones matemáticas, como pi, sin, cos, etc.
# Se importa para usar math.pi en el cálculo del área y perímetro del círculo.

class Figura:
    """Clase base abstracta para figuras."""
    # una clase abstracta es aquella que no se instancia directamente, sino que sirve como plantilla para otras clases que la heredan,
    # en este caso Figura es una clase abstracta porque no tiene una implementación concreta de los métodos area() y perimetro(),
    # sino que las subclases deben implementarlos.

    def __init__(self, color="blanco"):
        self.color = color

    def area(self):
        raise NotImplementedError("Las subclases deben implementar area()")
        # raise NotImplementedError es una forma de indicar que este método debe ser implementado por las subclases,
        # si se llama directamente desde Figura, se lanzará esta excepción para indicar que no tiene una implementación concreta.

    def perimetro(self):
        raise NotImplementedError("Las subclases deben implementar perimetro()")

    def describir(self):
        """Método que usa los métodos polimórficos."""
        print(f"  {type(self).__name__:12} | Color: {self.color:8} | "
              f"Área: {self.area():8.2f} | Perímetro: {self.perimetro():.2f}")


class Circulo(Figura):
    def __init__(self, radio, color="blanco"):
        super().__init__(color)
        self.radio = radio

    def area(self):
        return math.pi * self.radio ** 2

    def perimetro(self):
        return 2 * math.pi * self.radio


class Rectangulo(Figura):
    def __init__(self, ancho, alto, color="blanco"):
        super().__init__(color)
        self.ancho = ancho
        self.alto  = alto

    def area(self):
        return self.ancho * self.alto

    def perimetro(self):
        return 2 * (self.ancho + self.alto)


class Triangulo(Figura):
    def __init__(self, base, altura, lado_a, lado_b, lado_c, color="blanco"):
        super().__init__(color)
        self.base    = base
        self.altura  = altura
        self.lado_a  = lado_a
        self.lado_b  = lado_b
        self.lado_c  = lado_c

    def area(self):
        return (self.base * self.altura) / 2

    def perimetro(self):
        return self.lado_a + self.lado_b + self.lado_c


class Cuadrado(Rectangulo):
    """El cuadrado es un rectángulo especial."""
    def __init__(self, lado, color="blanco"):
        super().__init__(lado, lado, color)  # ancho = alto = lado


# ── POLIMORFISMO EN ACCIÓN ─────────────────────────────────
figuras = [
    Circulo(5,"rojo"),
    Rectangulo(4, 6,"azul"),
    Triangulo(3, 4, 3, 4, 5,"verde"),
    Cuadrado(7,"amarillo"),
]

print("Reporte de figuras geométricas:")
print("-" * 70)
print(f"  {'Figura':12} | {'Color':10} | {'Área':12} | {'Perímetro'}")
print("-" * 70)

# El mismo método 'describir()' → comportamientos diferentes según la figura
for figura in figuras:
    figura.describir()

# Función que funciona con CUALQUIER figura (polimorfismo)
def figura_mas_grande(lista_figuras):
    return max(lista_figuras, key=lambda f: f.area())
#max es una función incorporada en Python que devuelve el elemento máximo de un iterable o el máximo de dos o más argumentos,
# en este caso se usa para encontrar la figura con el área más grande en la lista de figuras, utilizando una función lambda como
# clave para comparar las áreas de las figuras.

#key=lambda f: f.area() es una función anónima que toma un objeto 'f' (que representa una figura) y devuelve el resultado de llamar al método area()
# de esa figura, se utiliza como criterio para determinar cuál es la figura con el área más grande al usar la función max().

mayor = figura_mas_grande(figuras)
print(f"\nFigura con mayor área: {type(mayor).__name__} ({mayor.area():.2f} u²)")
````

````python
# ============================================================
# EJEMPLO 8: Polimorfismo con Duck Typing
# ============================================================
# "Si camina como un pato y grazna como un pato, es un pato"
# Python no requiere herencia formal para el polimorfismo

class Pato:
    def hablar(self):
        return "¡Cuac cuac!"

class Persona:
    def hablar(self):
        return "¡Hola, buenos días!"

class Robot:
    def hablar(self):
        return "01001000 01101111 01101100 01100001"

class Perrito:
    def hablar(self):
        return "¡Guau!"


# Esta función funciona con CUALQUIER objeto que tenga método 'hablar()'
# No importa si son parientes o no
def hacer_hablar(ser):
    tipo = type(ser).__name__
    print(f"  {tipo:10}: {ser.hablar()}")


print("Demostración de Duck Typing:")
print("-" * 40)

seres = [Pato(), Persona(), Robot(), Perrito()]
#seres es una lista que contiene instancias de diferentes clases (Pato, Persona, Robot, Perrito),
# cada una con su propia implementación del método hablar(), lo que permite demostrar el concepto de Duck Typing en Python,
# donde el tipo específico del objeto no importa mientras tenga el método requerido (hablar() en este caso).

for ser in seres:
    hacer_hablar(ser)  # ¡Mismo código, distintos comportamientos!
````

## 7. Abstracción

### 📖 Concepto

La **abstracción** consiste en mostrar solo los detalles esenciales y ocultar la complejidad interna. Se enfoca en *qué hace* un objeto, no en *cómo lo hace*.

En Python se implementa con **clases abstractas** del módulo `abc` (Abstract Base Classes).

### 🔑 Características

- Una **clase abstracta** no puede instanciarse directamente
- Define métodos abstractos que las subclases **obligatoriamente** deben implementar
- Actúa como un *contrato* o *interfaz*

````python
# ============================================================
# EJEMPLO 9: Abstracción con ABC
# ============================================================

from abc import ABC, abstractmethod

class VehiculoAbstracto(ABC):
    """Clase abstracta: define el contrato para todos los vehículos."""

    def __init__(self, marca, modelo, año):
        self.marca  = marca
        self.modelo = modelo
        self.año    = año
        self._velocidad_actual = 0

    # ── MÉTODOS ABSTRACTOS ─────────────────────────────────
    # Las subclases DEBEN implementarlos o dará error
    @abstractmethod
    # El decorador @abstractmethod se utiliza para marcar un método como abstracto, lo que significa que no tiene una implementación concreta en la clase base (en este caso, VehiculoAbstracto) y debe ser implementado por cualquier clase que herede de ella. Si una subclase no implementa este método, se producirá un error al intentar instanciarla, asegurando así que todas las clases derivadas cumplan con el contrato definido por la clase abstracta.
    def arrancar(self):
        """Cómo arranca este vehículo."""
        pass

    @abstractmethod
    def tipo_combustible(self):
        """Qué tipo de combustible usa."""
        pass
    # El uso de 'pass' aquí es solo para indicar que no hay implementación en la clase abstracta, ya que se espera que las subclases proporcionen
    # su propia implementación de este método. Es una convención común en Python para métodos abstractos o vacíos, aunque no es estrictamente
    # necesario si se lanza una excepción como NotImplementedError.

    @abstractmethod
    def velocidad_maxima(self):
        """Velocidad máxima en km/h."""
        pass

    # ── MÉTODOS CONCRETOS (implementados en la base) ───────
    def frenar(self):
        self._velocidad_actual = 0
        print(f"{self.marca} {self.modelo} ha frenado.")

    def mostrar_info(self):
        print(f"\n🚗 {self.marca} {self.modelo} ({self.año})")
        print(f"   Combustible: {self.tipo_combustible()}")
        print(f"   Vel. máxima: {self.velocidad_maxima()} km/h")


# ── Implementaciones concretas ─────────────────────────────

class AutoElectrico(VehiculoAbstracto):
    def __init__(self, marca, modelo, año, capacidad_bateria):
        super().__init__(marca, modelo, año)
        self.capacidad_bateria = capacidad_bateria  # kWh

    def arrancar(self):  # Implementación obligatoria
        print(f"⚡ {self.marca} {self.modelo}: modo silencioso activado...")

    def tipo_combustible(self):
        return f"Eléctrico ({self.capacidad_bateria} kWh)"

    def velocidad_maxima(self):
        return 250

    def cargar_bateria(self):
        print(f"🔌 Cargando batería de {self.capacidad_bateria} kWh...")


class AutoGasolina(VehiculoAbstracto):
    def __init__(self, marca, modelo, año, cilindrada):
        super().__init__(marca, modelo, año)
        self.cilindrada = cilindrada

    def arrancar(self):
        print(f"🔑 {self.marca} {self.modelo}: ¡Vroooom! Motor encendido.")

    def tipo_combustible(self):
        return f"Gasolina ({self.cilindrada}cc)"

    def velocidad_maxima(self):
        return 200


class Motocicleta(VehiculoAbstracto):
    def __init__(self, marca, modelo, año):
        super().__init__(marca, modelo, año)

    def arrancar(self):
        print(f"🏍️ {self.marca} {self.modelo}: ¡Rum rum rum!")

    def tipo_combustible(self):
        return "Gasolina (moto)"

    def velocidad_maxima(self):
        return 300


# Intentar instanciar la clase abstracta → ERROR
print("Intentando crear VehiculoAbstracto directamente:")
try:
    v = VehiculoAbstracto("X", "Y", 2020)
except TypeError as e:
    # TypeError es una clase de excepción que se lanza cuando se intenta realizar una operación o función con un tipo de dato inapropiado,
    # en este caso se lanza porque VehiculoAbstracto es una clase abstracta y no se puede instanciar directamente.
    print(f"❌ Error correcto: {e}")

# Creando instancias concretas
print("\nCreando vehículos concretos:")
tesla    = AutoElectrico("Tesla",   "Model 3",  2023, 75)
toyota   = AutoGasolina("Toyota",  "Corolla",   2022, 1800)
yamaha   = Motocicleta("Yamaha",   "R1",        2023)

vehiculos = [tesla, toyota, yamaha]

for v in vehiculos:
    v.arrancar()
    v.mostrar_info()
````

````python
# ============================================================
# SOLUCIÓN EJERCICIO 8.1
# ============================================================

from abc import ABC, abstractmethod

class Empleado(ABC):
    def __init__(self, nombre, id_empleado, salario_base):
        self.nombre       = nombre
        self.id_empleado  = id_empleado
        self.salario_base = salario_base

    @abstractmethod
    def calcular_pago(self):
        pass

    def __str__(self):
        return f"[{type(self).__name__}] {self.nombre} (ID: {self.id_empleado})"


class EmpleadoPorHoras(Empleado):
    def __init__(self, nombre, id_empleado, tarifa_hora, horas_trabajadas):
        super().__init__(nombre, id_empleado, tarifa_hora)
        self.horas_trabajadas = horas_trabajadas

    def calcular_pago(self):
        pago = self.salario_base * self.horas_trabajadas
        # Horas extra (más de 40) se pagan al 150%
        if self.horas_trabajadas > 40:
            horas_extra = self.horas_trabajadas - 40
            pago = (40 * self.salario_base) + (horas_extra * self.salario_base * 1.5)
        return pago


class EmpleadoAsalariado(Empleado):
    def __init__(self, nombre, id_empleado, salario_mensual):
        super().__init__(nombre, id_empleado, salario_mensual)

    def calcular_pago(self):
        return self.salario_base


class EmpleadoConComision(Empleado):
    def __init__(self, nombre, id_empleado, salario_base, ventas_totales, tasa_comision):
        super().__init__(nombre, id_empleado, salario_base)
        self.ventas_totales  = ventas_totales
        self.tasa_comision   = tasa_comision

    def calcular_pago(self):
        return self.salario_base + (self.ventas_totales * self.tasa_comision)


def procesar_nomina(empleados):
    print("\n" + "=" * 50)
    print(" 💼 PROCESAMIENTO DE NÓMINA")
    print("=" * 50)
    total_empresa = 0
    for emp in empleados:
        pago = emp.calcular_pago()
        total_empresa += pago
        print(f"  {str(emp):40} | Pago: ${pago:,.2f}")
    print("-" * 50)
    print(f"  {'TOTAL NÓMINA':40} | ${total_empresa:,.2f}")
    print("=" * 50)


equipo = [
    EmpleadoPorHoras("Ana Torres",    1, 15.0,  45),     # 40 normales + 5 extra
    EmpleadoAsalariado("Bob Díaz",    2, 3000),
    EmpleadoConComision("Eva Ruiz",   3, 1000,  5000, 0.1),
    EmpleadoPorHoras("Luis Paz",      4, 20.0,  38),
]

procesar_nomina(equipo)
````

### 🏋️ Ejercicio 8.2: Inventario de Productos

Crea un sistema de inventario con:
- Clase `Producto` con encapsulamiento del precio y stock
- Validación: precio no puede ser negativo, stock no puede ser negativo
- Métodos: `vender(cantidad)`, `reabastecer(cantidad)`, `aplicar_descuento(porcentaje)`
- Clase `Inventario` que gestione una colección de productos
- Métodos de Inventario: `agregar_producto()`, `buscar_por_nombre()`, `productos_agotados()`, `valor_total_inventario()`

````python
# ============================================================
# SOLUCIÓN EJERCICIO 8.2
# ============================================================

class Producto:
    def __init__(self, nombre, precio, stock, categoria):
        self.nombre     = nombre
        self.__precio   = precio  # Privado
        self.__stock    = stock   # Privado
        self.categoria  = categoria

    @property
    def precio(self):
        return self.__precio

    @precio.setter
    def precio(self, valor):
        if valor < 0:
            raise ValueError("El precio no puede ser negativo.")
        self.__precio = valor

    @property
    def stock(self):
        return self.__stock

    def vender(self, cantidad):
        if cantidad <= 0:
            raise ValueError("La cantidad debe ser positiva.")
        if cantidad > self.__stock:
            print(f"❌ Stock insuficiente de '{self.nombre}'. Disponible: {self.__stock}")
            return False
        self.__stock -= cantidad
        print(f"✅ Vendidas {cantidad} unidades de '{self.nombre}'. Stock: {self.__stock}")
        return True

    def reabastecer(self, cantidad):
        if cantidad <= 0:
            raise ValueError("La cantidad debe ser positiva.")
        self.__stock += cantidad
        print(f"📦 Reabastecidas {cantidad} unidades de '{self.nombre}'. Stock: {self.__stock}")

    def aplicar_descuento(self, porcentaje):
        if not 0 < porcentaje < 100:
            raise ValueError("El porcentaje debe estar entre 0 y 100.")
        descuento      = self.__precio * (porcentaje / 100)
        precio_anterior = self.__precio
        self.__precio   -= descuento
        print(f"🏷️  Descuento {porcentaje}% en '{self.nombre}': ${precio_anterior:.2f} → ${self.__precio:.2f}")

    def valor_en_inventario(self):
        return self.__precio * self.__stock

    def __str__(self):
        return (f"{self.nombre:20} | Cat: {self.categoria:12} | "
                f"Precio: ${self.__precio:8.2f} | Stock: {self.__stock:5}")


class Inventario:
    def __init__(self, nombre_tienda):
        self.nombre_tienda = nombre_tienda
        self.__productos   = {}  # Diccionario: nombre → Producto

    def agregar_producto(self, producto):
        self.__productos[producto.nombre] = producto
        print(f"➕ Producto '{producto.nombre}' agregado al inventario.")

    def buscar_por_nombre(self, nombre):
        return self.__productos.get(nombre, None)

    def productos_agotados(self):
        return [p for p in self.__productos.values() if p.stock == 0]

    def valor_total_inventario(self):
        return sum(p.valor_en_inventario() for p in self.__productos.values())

    def mostrar_catalogo(self):
        print(f"\n{'='*70}")
        print(f"  🏪 INVENTARIO: {self.nombre_tienda}")
        print(f"{'='*70}")
        for p in self.__productos.values():
            estado = "🔴 AGOTADO" if p.stock == 0 else ("🟡 BAJO" if p.stock < 5 else "🟢")
            print(f"  {p}  {estado}")
        print(f"{'─'*70}")
        print(f"  Valor total del inventario: ${self.valor_total_inventario():,.2f}")
        print(f"{'='*70}")


# ── DEMO ──────────────────────────────────────────────────
tienda = Inventario("TechStore Argentina")

p1 = Producto("Laptop HP",        1200.00, 10, "Electrónica")
p2 = Producto("Mouse Logitech",     25.00, 50, "Periféricos")
p3 = Producto("Teclado Mecánico",   80.00,  3, "Periféricos")
p4 = Producto("Monitor 27'",       350.00,  0, "Electrónica")

for p in [p1, p2, p3, p4]:
    tienda.agregar_producto(p)

tienda.mostrar_catalogo()

print("\nOperaciones:")
p1.vender(3)
p2.aplicar_descuento(20)
p3.reabastecer(10)
p4.vender(1)  # Intento fallido

tienda.mostrar_catalogo()

agotados = tienda.productos_agotados()
print(f"\nProductos agotados: {[p.nombre for p in agotados]}")
````

## 9. Proyecto Final: Sistema de Biblioteca

### 📖 Descripción

Desarrollaremos un sistema completo de biblioteca que integra **todos** los pilares de la POO:

- **Abstracción**: Clase abstracta `MaterialBiblioteca`
- **Herencia**: `Libro`, `Revista`, `DVD` heredan de `MaterialBiblioteca`
- **Encapsulamiento**: Stock y estado de préstamo protegidos
- **Polimorfismo**: Todos los materiales comparten el método `mostrar_info()`

````python
# ============================================================
# PROYECTO FINAL: Sistema de Biblioteca
# ============================================================

from abc import ABC, abstractmethod
from datetime import datetime, timedelta


# ── ABSTRACCIÓN ────────────────────────────────────────────
class MaterialBiblioteca(ABC):
    """Clase abstracta base para todos los materiales."""

    def __init__(self, titulo, año, codigo):
        self.titulo     = titulo
        self.año        = año
        self.codigo     = codigo
        self.__prestado = False
        self.__fecha_devolucion = None

    @property
    def prestado(self):
        return self.__prestado

    @property
    def fecha_devolucion(self):
        return self.__fecha_devolucion

    def prestar(self, dias=14):
        """Presta el material por N días."""
        if self.__prestado:
            print(f"❌ '{self.titulo}' ya está prestado hasta {self.__fecha_devolucion.strftime('%d/%m/%Y')}.")
            return False
        self.__prestado         = True
        self.__fecha_devolucion = datetime.now() + timedelta(days=dias)
        print(f"✅ '{self.titulo}' prestado. Devolver antes del {self.__fecha_devolucion.strftime('%d/%m/%Y')}.")
        return True

    def devolver(self):
        """Devuelve el material."""
        if not self.__prestado:
            print(f"⚠️  '{self.titulo}' no estaba prestado.")
            return
        self.__prestado         = False
        self.__fecha_devolucion = None
        print(f"📚 '{self.titulo}' devuelto. ¡Gracias!")

    @abstractmethod
    def mostrar_info(self):
        """Muestra información específica del material."""
        pass

    @abstractmethod
    def tipo(self):
        """Retorna el tipo de material."""
        pass

    def estado(self):
        return "🔴 Prestado" if self.__prestado else "🟢 Disponible"


# ── HERENCIA ───────────────────────────────────────────────
class Libro(MaterialBiblioteca):
    def __init__(self, titulo, autor, año, codigo, isbn, paginas, genero):
        super().__init__(titulo, año, codigo)
        self.autor   = autor
        self.isbn    = isbn
        self.paginas = paginas
        self.genero  = genero

    def tipo(self):
        return "📖 Libro"

    def mostrar_info(self):
        print(f"  {self.tipo()} | Código: {self.codigo}")
        print(f"    Título:  {self.titulo}")
        print(f"    Autor:   {self.autor}")
        print(f"    Año:     {self.año} | Páginas: {self.paginas} | Género: {self.genero}")
        print(f"    Estado:  {self.estado()}")


class Revista(MaterialBiblioteca):
    def __init__(self, titulo, editorial, año, codigo, numero, tematica):
        super().__init__(titulo, año, codigo)
        self.editorial = editorial
        self.numero    = numero
        self.tematica  = tematica

    def tipo(self):
        return "📰 Revista"

    def mostrar_info(self):
        print(f"  {self.tipo()} | Código: {self.codigo}")
        print(f"    Título:    {self.titulo} N°{self.numero}")
        print(f"    Editorial: {self.editorial} | Temática: {self.tematica}")
        print(f"    Estado:    {self.estado()}")


class DVD(MaterialBiblioteca):
    def __init__(self, titulo, director, año, codigo, duracion_min, clasificacion):
        super().__init__(titulo, año, codigo)
        self.director      = director
        self.duracion_min  = duracion_min
        self.clasificacion = clasificacion

    def tipo(self):
        return "🎬 DVD"

    def mostrar_info(self):
        print(f"  {self.tipo()} | Código: {self.codigo}")
        print(f"    Título:   {self.titulo}")
        print(f"    Director: {self.director} | Duración: {self.duracion_min} min | Clasificación: {self.clasificacion}")
        print(f"    Estado:   {self.estado()}")


# ── CLASE GESTORA ──────────────────────────────────────────
class Biblioteca:
    def __init__(self, nombre, ciudad):
        self.nombre     = nombre
        self.ciudad     = ciudad
        self.__catalogo = {}  # codigo → MaterialBiblioteca

    def registrar(self, material):
        self.__catalogo[material.codigo] = material

    def buscar(self, codigo):
        return self.__catalogo.get(codigo)

    def buscar_por_titulo(self, titulo):
        return [m for m in self.__catalogo.values()
                if titulo.lower() in m.titulo.lower()]

    def disponibles(self):
        return [m for m in self.__catalogo.values() if not m.prestado]

    def prestados(self):
        return [m for m in self.__catalogo.values() if m.prestado]

    def mostrar_catalogo_completo(self):
        print(f"\n{'='*60}")
        print(f"  🏛️  BIBLIOTECA: {self.nombre} – {self.ciudad}")
        print(f"  Total de materiales: {len(self.__catalogo)}")
        print(f"  Disponibles: {len(self.disponibles())} | Prestados: {len(self.prestados())}")
        print(f"{'='*60}")
        for material in self.__catalogo.values():
            material.mostrar_info()  # ← POLIMORFISMO
            print()


# ── DEMO COMPLETA ──────────────────────────────────────────
print("Iniciando Sistema de Biblioteca...\n")

biblio = Biblioteca("Biblioteca Central", "Posadas")

# Registrando materiales diversos
libros = [
    Libro("Cien Años de Soledad", "G. García Márquez", 1967, "L001", "978-84-376-0494-7", 471, "Realismo mágico"),
    Libro("El Principito",        "Antoine de Saint-Exupéry", 1943, "L002", "978-84-261-3444-9", 96,  "Ficción"),
    Libro("Python Fluente",       "Luciano Ramalho",   2015, "L003", "978-1-491-94600-8", 792, "Tecnología"),
]

revistas = [
    Revista("National Geographic", "Nat Geo Soc.", 2024, "R001", 350, "Ciencia y Naturaleza"),
    Revista("IEEE Spectrum",       "IEEE",          2024, "R002", 612, "Tecnología"),
]

dvds = [
    DVD("Inception",        "Christopher Nolan", 2010, "D001", 148, "ATP"),
    DVD("Parasite",         "Bong Joon-Ho",      2019, "D002", 132, "ATP"),
]

for item in libros + revistas + dvds:
    biblio.registrar(item)

biblio.mostrar_catalogo_completo()

# Operaciones de préstamo
print("\n" + "=" * 60)
print("  OPERACIONES DE PRÉSTAMO")
print("=" * 60)

biblio.buscar("L001").prestar()
biblio.buscar("D001").prestar(7)
biblio.buscar("L001").prestar()  # Intento fallido

print(f"\nMateriales actualmente prestados:")
for m in biblio.prestados():
    print(f"  • {m.tipo()} '{m.titulo}' – devolver: {m.fecha_devolucion.strftime('%d/%m/%Y')}")

print("\nDevolviendo materiales:")
biblio.buscar("L001").devolver()

# Búsqueda por título
print("\nBúsqueda de 'Python':")
resultados = biblio.buscar_por_titulo("python")
for r in resultados:
    print(f"  Encontrado: {r.titulo}")

print("\n✅ Sistema de Biblioteca funcionando correctamente.")
````

## 📊 Resumen Final: Los 4 Pilares de la POO

| Pilar | Definición | Herramienta en Python | Beneficio |
|-------|-----------|----------------------|----------|
| **Encapsulamiento** | Ocultar detalles internos | `__atributo`, `@property` | Protege los datos, facilita validación |
| **Herencia** | Reutilizar y extender clases | `class Hija(Padre)`, `super()` | Reutilización de código, jerarquías naturales |
| **Polimorfismo** | Mismo método, distintos comportamientos | Sobreescritura de métodos, duck typing | Código flexible y extensible |
| **Abstracción** | Mostrar solo lo esencial | `ABC`, `@abstractmethod` | Define contratos, simplifica uso |
