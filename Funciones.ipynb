## Funciones

Las funciones nos permiten reutilizar y organizar el código. Por ejemplo, digamos que tenemos un código que calcula el área de un círculo:

```python
radius = 5
area = 3.14 * radius * radius
```

¡Eso funciona! El problema es cuando queremos calcular el área de otros círculos, cada uno con su propio radio. Podríamos simplemente copiar el código y cambiar los nombres de las variables de esta manera:

```python
radius = 5
area1 = 3.14 * radius * radius

radius2 = 7
area2 = 3.14 * radius2 * radius2
```

¡Pero queremos reutilizar nuestro código! ¿Por qué querríamos rehacer nuestro trabajo? ¿Qué pasa si queremos calcular el área de miles de círculos? Ahí es donde las funciones ayudan.
En su lugar, podemos definir una nueva función llamada area_of_circle usando la palabra clave def.

```python
def area_of_circle(r):
    pi = 3.14
    result = pi * r * r
    return result
```

Analicemos esta función area_of_circle:

Requiere una entrada (también conocida como "parámetro" o "argumento") llamada r.

El cuerpo de la función está identado: este es el código que se ejecutará cada vez que usemos la función.

Devuelve un único valor (la salida de la función). En este caso, es el valor almacenado en la variable result.

Para "llamar" a esta función podemos poner cualquier número como argumento (en este caso, 5) para el parámetro r y capturar la salida en una nueva variable:

```python
area = area_of_circle(5)
print(area)
```

1 - 5 es la entrada o argumento que le dimos a r

2 - Se ejecuta el cuerpo de la función, que almacena 78,5 en la variable result.

3 - La función devuelve el valor 78,5, lo que significa que la expresión area_of_circle(5) se evalúa como 78,5.

4 - 78,5 se almacena en la variable area y luego se imprime

Como ya hemos definido la función, ¡ahora podemos usarla tantas veces como queramos con diferentes entradas!

```python
area = area_of_circle(6)
print(area)
# 113.04

area = area_of_circle(7)
print(area)
# 153.86
```

## Ejemplo con ejercicio
We need to calculate the size of a weapon's "attack area". With a 1.0 meter sword, for example, a player can attack in an area of 3.14 square meters around them. You can use the area_of_circle function to do that calculation.



```python
def area_of_circle(radius):
    pi = 3.14
    area = pi * radius * radius
    return area


sword_length = 1.0
spear_length = 2.0


sword_area = area_of_circle(sword_length)
spear_area = area_of_circle(spear_length)


print("Sword length:", sword_length, "meters.")
print("Sword attack area:", sword_area, "square meters")

print("Spear length:", spear_length, "meters.")
print("Spear attack area:", spear_area, "square meters")
```

## Printing vs Returning

Puede resultar especialmente confuso cuando el conjunto de pruebas que proporcionamos simplemente imprime la salida de sus funciones en la consola. Parece que print() y return son intercambiables, ¡pero no lo son!

**Print()** es una función que:

Imprime un valor en la consola

No devuelve un valor

**Return** es una palabra clave que:

Finaliza la ejecución de la función actual.

Proporciona un valor (o valores) al llamador de la función.

No imprime nada en la consola (a menos que el valor de retorno se imprima posteriormente).



## Orden de funciones
Todas las funciones deben definirse antes de ser utilizadas.

Puede que pienses que esto dificultaría la estructuración del código Python porque el orden de las funciones debe ser el correcto. Resulta que hay un truco simple que lo hace muy fácil.

La mayoría de los desarrolladores de Python resuelven este problema definiendo primero todas las funciones en su programa y luego llaman a una función de "punto de entrada" por último. De esa manera, el intérprete de Python ha leído todas las funciones antes de llamar a la primera.

Nota: convencionalmente, esta función de "punto de entrada" suele llamarse **main** para mantener las cosas simples y consistentes.

```python
def main():
    health = 10
    armor = 5
    add_armor(health, armor)

def add_armor(h, a):
    new_health = h + a
    print_health(new_health)

def print_health(new_health):
    print(f"The player now has {new_health} health")

# call entrypoint last
main()
```

## Parámetros vs argumentos
Los parámetros son los nombres que se usan para las entradas cuando se define una función. Los argumentos son los valores de las entradas que se proporcionan cuando se llama a una función.
Para reiterar, los argumentos son los valores reales que se incluyen en la función, como 42.0, "el caballero oscuro" o True. Los parámetros son los nombres que usamos en la definición de la función para referirnos a esos valores, que al momento de escribir la función, pueden ser los que queramos.

```python
# a y b son parametros
def add(a, b):
    return a + b

# 5 y 6 son argumentos
sum = add(5, 6)
```

### Funciones que no devuelven nada vs funciones que sí


Hay dos tipos de funciones: las que HACEN algo (imprimir, guardar un archivo) y las que CALCULAN algo (y devuelven el resultado con return). Las primeras son como un robot que mueve cajas — hace un trabajo pero no te da nada. Las segundas son como una calculadora — le das números y te devuelve un resultado que puedes guardar en una variable.

```python
# Función que HACE algo (no return)
def mostrar_ticket(producto, precio, cantidad):
    """Imprime un ticket de compra formateado."""
    total = precio * cantidad
    print("=" * 30)
    print(f"Producto: {producto}")
    print(f"Precio unidad: {precio:.2f}€")
    print(f"Cantidad: {cantidad}")
    print(f"TOTAL: {total:.2f}€")
    print("=" * 30)

mostrar_ticket("Café", 1.50, 3)

# Función que CALCULA algo (con return)
def calcular_media(numeros):
    """Calcula la media de una lista de números."""
    suma = 0
    for n in numeros:
        suma = suma + n
    media = suma / len(numeros)
    return media

notas = [7.5, 8.0, 6.5, 9.0, 7.0]
mi_media = calcular_media(notas)
print(f"Tu media es: {mi_media}")  # 7.6

# Puedes usar el resultado en decisiones
if mi_media >= 7:
    print("¡Notable!")
```

### Explicacion detallada

Línea por línea:

- def mostrar_ticket(...): → esta función HACE algo (imprime un ticket) pero NO devuelve nada. Es como un robot que coloca carteles: trabaja pero no te da un resultado numérico.
- mostrar_ticket("Café", 1.50, 3) → la llamas y ella imprime. Pero no puedes guardar su resultado en una variable (sería None).
- def calcular_media(numeros): → esta función CALCULA algo y lo devuelve con return. Es como una calculadora: le das datos y te devuelve un número.
- return media → "aquí tienes el resultado, haz con él lo que quieras".
- mi_media = calcular_media(notas) → guardas el valor devuelto en una variable. Ahora puedes usarlo en un if, en otro cálculo, etc.

La diferencia clave: sin return, la función hace un efecto secundario (imprimir). Con return, la función produce un valor que puedes capturar. Ambas son útiles, pero las funciones con return son más versátiles porque puedes REUTILIZAR su resultado.

### Resumen
▹ def nombre(parámetros): define una función.

▹ return devuelve un valor al código que llamó a la función.

▹ Una función sin return implícitamente devuelve None.

▹ Los parámetros son los datos que la función necesita para trabajar.

▹ Las variables dentro de la función son locales (no existen fuera).

▹ Una función debe hacer UNA cosa y hacerla bien.

▹ Puedes combinar funciones llamando unas desde otras.

## Ejercicio Validador de Contrasenas

```python
passwords = ["Python2024", "Sol3", "password123", "Password", "Python12"]

def tiene_longitud_minima(password):
    return len(password) >= 8

def tiene_numero(password):
    for char in password:
        if char.isdigit():
            return True
    return False

def tiene_mayuscula(password):
    for char in password:
        if char.isupper():
            return True
    return False

def es_password_segura(password):
    return tiene_longitud_minima(password) and tiene_numero(password) and tiene_mayuscula(password)

# No toques nada de aquí abajo: es lo que comprueba tus funciones
for pw in passwords:
    if es_password_segura(pw):
        print(f"{pw:14} ✅ segura")
    else:
        motivos = ""
        if not tiene_longitud_minima(pw):
            motivos = motivos + " corta"
        if not tiene_numero(pw):
            motivos = motivos + " sin-número"
        if not tiene_mayuscula(pw):
            motivos = motivos + " sin-mayúscula"
        print(f"{pw:14} ❌{motivos}")
```
