### **Objetivo de la Actividad**

Comprender y calcular la probabilidad condicional, y diferenciar entre eventos independientes y mutuamente exclusivos utilizando un conjunto de datos real.

-----

### **Paso 1: Preparación y Carga de Datos**

En este paso, simularás un dataset simple de vuelos para un análisis. Este conjunto de datos te permitirá analizar las relaciones entre dos variables categóricas: **clima** y **retraso de vuelo**.

1.  **Instalar librerías:**

    ```bash
    pip install pandas numpy matplotlib
    ```

2.  **Crear el DataFrame:**

````
import pandas as pd
import numpy as np

# Simulación de datos de 2000 vuelos
np.random.seed(42)
datos = {
    'Clima': np.random.choice(['Soleado', 'Lluvioso'], size=2000, p=[0.7, 0.3]),
    'Retraso_Vuelo': np.random.choice(['Sí', 'No'], size=2000, p=[0.2, 0.8])
}
df = pd.DataFrame(datos)

# Introducir un sesgo para que la lluvia cause más retrasos
df.loc[df['Clima'] == 'Lluvioso', 'Retraso_Vuelo'] = np.random.choice(['Sí', 'No'], size=(df['Clima'] == 'Lluvioso').sum(), p=[0.6, 0.4])

print("Primeros 5 registros del DataFrame:")
print(df.head())
````

-----

### **Paso 2: Medidas de Probabilidad Básica**

Antes de pasar a la probabilidad condicional, es útil calcular las probabilidades simples de cada evento.

1.  **Probabilidad de cada evento:**

      * `P(A)`: Probabilidad de que el clima sea **Lluvioso**.
      * `P(B)`: Probabilidad de que haya un **Retraso de Vuelo**.

    <!-- end list -->

````
# Usamos pandas para contar las frecuencias
prob_lluvia = (df['Clima'] == 'Lluvioso').mean()
prob_retraso = (df['Retraso_Vuelo'] == 'Sí').mean()

print("\n--- Probabilidades Básicas ---")
print(f"Probabilidad de que el clima sea Lluvioso (P(Lluvia)): {prob_lluvia:.2f}")
print(f"Probabilidad de que haya un Retraso (P(Retraso)): {prob_retraso:.2f}")
````

-----

### **Paso 3: Probabilidad Condicional**

La **probabilidad condicional** es la probabilidad de que un evento ocurra **dado que otro evento ya ha ocurrido**. Se denota como `P(A|B)`, que se lee "la probabilidad de A dado B".

1.  **Calcula `P(Retraso | Lluvioso)`:**

      * **Consigna:** Calcula la probabilidad de que un vuelo se retrase **dado que el clima es lluvioso**.

    <!-- end list -->

````
# Filtra los datos para considerar solo los vuelos en días de lluvia
df_lluvia = df[df['Clima'] == 'Lluvioso']

# Dentro de ese subconjunto, calcula la probabilidad de retraso
prob_retraso_dado_lluvia = (df_lluvia['Retraso_Vuelo'] == 'Sí').mean()

print("\n--- Probabilidad Condicional ---")
print(f"Probabilidad de Retraso dado Lluvia (P(Retraso | Lluvia)): {prob_retraso_dado_lluvia:.2f}")

print("\nConclusión:")
print(f"La probabilidad de un retraso de vuelo es del {prob_retraso*100:.0f}%, pero si sabemos que el clima es lluvioso, la probabilidad aumenta a {prob_retraso_dado_lluvia*100:.0f}%.")
````

-----

### **Paso 4: Eventos Independientes y Mutuamente Exclusivos**

  * **Eventos Independientes:** Dos eventos son independientes si la ocurrencia de uno no afecta la probabilidad del otro. Se cumple si `P(A|B) = P(A)`.
  * **Eventos Mutuamente Exclusivos:** Dos eventos son mutuamente exclusivos (o disjuntos) si no pueden ocurrir al mismo tiempo. Por ejemplo, no puedes ser un vuelo retrasado y no-retrasado simultáneamente.

<!-- end list -->

1.  **Prueba de Independencia:**

      * **Consigna:** Compara la probabilidad condicional `P(Retraso | Lluvioso)` con la probabilidad simple `P(Retraso)`.
      * **Pregunta a responder:** ¿Son los eventos "Lluvia" y "Retraso de Vuelo" independientes? ¿Por qué?

    <!-- end list -->

````
son_independientes = np.isclose(prob_retraso_dado_lluvia, prob_retraso, atol=0.05)
print("\n--- Análisis de Independencia ---")
print(f"P(Retraso | Lluvia) = {prob_retraso_dado_lluvia:.2f}")
print(f"P(Retraso) = {prob_retraso:.2f}")

print(son_independientes)

if son_independientes:
    print("Los eventos 'Lluvia' y 'Retraso de Vuelo' son independientes.")
else:
    print("Los eventos 'Lluvia' y 'Retraso de Vuelo' NO son independientes, ya que P(Retraso | Lluvia) != P(Retraso).")
print("La lluvia afecta la probabilidad de que un vuelo se retrase, por lo que no son eventos independientes.")
````

2.  **Prueba de Eventos Mutuamente Exclusivos:**

      * **Consigna:** ¿Son los eventos "Vuelo Retrasado" y "Vuelo No Retrasado" mutuamente exclusivos?
      * **Pregunta a responder:** ¿Pueden ocurrir al mismo tiempo?

    <!-- end list -->

````
print("\n--- Análisis de Eventos Mutuamente Exclusivos ---")
print("¿Son los eventos 'Vuelo Retrasado' y 'Vuelo No Retrasado' mutuamente exclusivos?")

# Calcular la intersección de ambos eventos
prob_retraso_y_no_retraso = ((df['Retraso_Vuelo'] == 'Sí') & (df['Retraso_Vuelo'] == 'No')).mean()

if prob_retraso_y_no_retraso == 0:
    print("Sí, son eventos mutuamente exclusivos, porque su intersección es 0. Un vuelo no puede estar retrasado y no retrasado al mismo tiempo.")
else:
    print("No, no son mutuamente exclusivos.")
````

### **Paso 5: Resumen y Conclusiones**

  * **Probabilidad Condicional:** Esta es la herramienta clave para el razonamiento probabilístico en el mundo real. Muestra cómo la información adicional puede cambiar drásticamente la probabilidad de un evento.
  * **Independencia:** Si la probabilidad condicional es diferente de la probabilidad simple, los eventos están relacionados. En este caso, el clima y los retrasos están claramente vinculados.
* **Eventos Mutuamente Exclusivos:** Son eventos que no tienen resultados en común. La clave es que su intersección es cero.
